```python
import numpy as np 
from scipy.stats import gaussian_kde
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/GSE138662_raw_counts_GRCh38.p13_NCBI.tsv', delimiter='\t')
data = df.iloc[:, 1:].to_numpy()
```

```python
vars = np.var(data, axis=1)
kde = gaussian_kde(vars)
x = np.linspace(np.min(vars), np.max(vars), 1000)
plt.plot(x, kde(x))
plt.show()
```

```python
idxs = np.argsort(vars)[::-1][:100]

top_vars = vars[idxs]
kde = gaussian_kde(top_vars)
x = np.linspace(np.min(top_vars), np.max(top_vars), 1000)
plt.plot(x, kde(x))
plt.show()
```
