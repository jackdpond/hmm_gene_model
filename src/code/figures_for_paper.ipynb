```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
fig, ax = plt.subplots(figsize=(18, 6), dpi=300)

x = np.linspace(-20, 20, 1000)
y = np.arcsinh(x)

ax.plot(x, y, linewidth=3)
ax.set(xlim=(-20, 20), ylim=(-5, 5), title="arcsinh(x)")
plt.show()
plt.savefig('arcsinh_figure.png')
```
